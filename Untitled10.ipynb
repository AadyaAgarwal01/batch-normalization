```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, random_split
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay

# Set device (GPU if available, otherwise CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Convert images to tensors
transform = transforms.ToTensor()

# Download full training data (60,000) and test data (10,000)
full_train_data = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_data = datasets.MNIST(root='./data', train=False, download=True, transform=transform)

# Split 60,000 training images into 50,000 Train and 10,000 Validation
train_data, val_data = random_split(full_train_data, [50000, 10000])

# Create mini-batches (Paper uses batch size = 60 in Section 4.1)
train_loader = DataLoader(train_data, batch_size=60, shuffle=True)
val_loader = DataLoader(val_data, batch_size=1000, shuffle=False)
test_loader = DataLoader(test_data, batch_size=1000, shuffle=False)

print(f"Train samples: {len(train_data)} | Validation samples: {len(val_data)} | Test samples: {len(test_data)}")
```

```python
class NetworkWithoutBN(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()
        self.net = nn.Sequential(
            nn.Linear(28 * 28, 100),
            nn.Sigmoid(),
            nn.Linear(100, 100),
            nn.Sigmoid(),
            nn.Linear(100, 100),
            nn.Sigmoid(),
            nn.Linear(100, 10)  # 10 output classes (digits 0 to 9)
        )

    def forward(self, x):
        x = self.flatten(x)
        return self.net(x)

model_without_bn = NetworkWithoutBN().to(device)
```

```python
class NetworkWithBN(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()
        self.net = nn.Sequential(
            nn.Linear(28 * 28, 100, bias=False),
            nn.BatchNorm1d(100),  # Added Batch Normalization
            nn.Sigmoid(),

            nn.Linear(100, 100, bias=False),
            nn.BatchNorm1d(100),  # Added Batch Normalization
            nn.Sigmoid(),

            nn.Linear(100, 100, bias=False),
            nn.BatchNorm1d(100),  # Added Batch Normalization
            nn.Sigmoid(),

            nn.Linear(100, 10)
        )

    def forward(self, x):
        x = self.flatten(x)
        return self.net(x)

model_with_bn = NetworkWithBN().to(device)
```

```python
def train_and_evaluate(model, lr, epochs=10):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.SGD(model.parameters(), lr=lr)

    history = {"train_loss": [], "val_acc": []}

    for epoch in range(epochs):
        # 1. Training Mode
        model.train()
        total_loss = 0
        for images, labels in train_loader:
            images, labels = images.to(device), labels.to(device)

            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        avg_loss = total_loss / len(train_loader)
        history["train_loss"].append(avg_loss)

        # 2. Validation Mode (Checked after every epoch)
        model.eval()
        val_preds, val_labels = [], []
        with torch.no_grad():
            for images, labels in val_loader:
                images = images.to(device)
                outputs = model(images)
                preds = torch.argmax(outputs, dim=1).cpu().numpy()
                val_preds.extend(preds)
                val_labels.extend(labels.numpy())

        val_acc = accuracy_score(val_labels, val_preds) * 100
        history["val_acc"].append(val_acc)
        print(f"Epoch [{epoch+1}/{epochs}] - Loss: {avg_loss:.4f} | Val Accuracy: {val_acc:.2f}%")

    # 3. Final Test Mode (Runs once on unseen test_loader after training finishes)
    model.eval()
    test_preds, test_labels = [], []
    with torch.no_grad():
        for images, labels in test_loader:
            images = images.to(device)
            outputs = model(images)
            preds = torch.argmax(outputs, dim=1).cpu().numpy()
            test_preds.extend(preds)
            test_labels.extend(labels.numpy())

    return history, test_labels, test_preds
```

```python
print("--- Training Model 1: WITHOUT Batch Normalization ---")
history_no_bn, y_true_no_bn, y_pred_no_bn = train_and_evaluate(model_without_bn, lr=0.1, epochs=10)

print("\n--- Training Model 2: WITH Batch Normalization ---")
history_bn, y_true_bn, y_pred_bn = train_and_evaluate(model_with_bn, lr=0.5, epochs=10)
```

```python
def get_metrics(y_true, y_pred):
    acc = accuracy_score(y_true, y_pred) * 100
    prec = precision_score(y_true, y_pred, average='macro') * 100
    rec = recall_score(y_true, y_pred, average='macro') * 100
    f1 = f1_score(y_true, y_pred, average='macro') * 100
    return acc, prec, rec, f1

metrics_no_bn = get_metrics(y_true_no_bn, y_pred_no_bn)
metrics_bn = get_metrics(y_true_bn, y_pred_bn)

print(f"{'Metric':<15} | {'Without BN (Before)':<20} | {'With BN (After)':<20}")
print("-" * 60)
print(f"{'Accuracy':<15} | {metrics_no_bn[0]:<19.2f}% | {metrics_bn[0]:<19.2f}%")
print(f"{'Precision':<15} | {metrics_no_bn[1]:<19.2f}% | {metrics_bn[1]:<19.2f}%")
print(f"{'Recall':<15} | {metrics_no_bn[2]:<19.2f}% | {metrics_bn[2]:<19.2f}%")
print(f"{'F1-Score':<15} | {metrics_no_bn[3]:<19.2f}% | {metrics_bn[3]:<19.2f}%")
```

```python
epochs_range = range(1, 11)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Graph 1: Validation Accuracy Over Epochs
axes[0, 0].plot(epochs_range, history_no_bn["val_acc"], 'r--o', label="Without BN (Before)")
axes[0, 0].plot(epochs_range, history_bn["val_acc"], 'g-s', label="With BN (After)")
axes[0, 0].set_title("Validation Accuracy Over Epochs (Higher is Better)")
axes[0, 0].set_xlabel("Epochs")
axes[0, 0].set_ylabel("Validation Accuracy (%)")
axes[0, 0].legend()
axes[0, 0].grid(True)

# Graph 2: Training Loss Over Epochs
axes[0, 1].plot(epochs_range, history_no_bn["train_loss"], 'r--o', label="Without BN (Before)")
axes[0, 1].plot(epochs_range, history_bn["train_loss"], 'g-s', label="With BN (After)")
axes[0, 1].set_title("Training Loss Over Epochs (Lower is Better)")
axes[0, 1].set_xlabel("Epochs")
axes[0, 1].set_ylabel("Loss")
axes[0, 1].legend()
axes[0, 1].grid(True)

# Graph 3: Bar Chart of Final Test Metrics (Accuracy, Precision, Recall, F1-Score)
metric_names = ["Accuracy", "Precision", "Recall", "F1-Score"]
x = range(len(metric_names))
width = 0.35

axes[1, 0].bar([i - width/2 for i in x], metrics_no_bn, width, label="Without BN", color="salmon")
axes[1, 0].bar([i + width/2 for i in x], metrics_bn, width, label="With BN", color="seagreen")
axes[1, 0].set_xticks(x)
axes[1, 0].set_xticklabels(metric_names)
axes[1, 0].set_ylim(0, 105)
axes[1, 0].set_title("Final Test Set Metrics Comparison")
axes[1, 0].set_ylabel("Score (%)")
axes[1, 0].legend()
axes[1, 0].grid(axis='y')

# Graph 4: Confusion Matrix on Test Set (With Batch Normalization)
cm = confusion_matrix(y_true_bn, y_pred_bn)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(ax=axes[1, 1], colorbar=False, cmap="Blues")
axes[1, 1].set_title("Test Confusion Matrix (With Batch Normalization)")

plt.tight_layout()
plt.show()
```
